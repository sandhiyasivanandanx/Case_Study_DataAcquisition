**Part B – Data Understanding**

**TASK 5**

```python
population = pd.read_csv('population.csv')
gdp = pd.read_excel('gdp.xlsx')
internet_users = pd.read_json('internet_users.json')
literacy = pd.read_xml('literacy_rate.xml')
```

```python
print("POPULATION DATASET")
print("Rows:", population.shape[0])
print("Columns:", population.shape[1])
print("Column names:", population.columns.tolist())
print("Data types:\n", population.dtypes)
print("Missing values:\n", population.isnull().sum())
print("Duplicate records:", population.duplicated().sum())

print("\nGDP DATASET")
print("Rows:", gdp.shape[0])
print("Columns:", gdp.shape[1])
print("Column names:", gdp.columns.tolist())
print("Data types:\n", gdp.dtypes)
print("Missing values:\n", gdp.isnull().sum())
print("Duplicate records:", gdp.duplicated().sum())

print("\nINTERNET USERS DATASET")
print("Rows:", internet_users.shape[0])
print("Columns:", internet_users.shape[1])
print("Column names:", internet_users.columns.tolist())
print("Data types:\n", internet_users.dtypes)
print("Missing values:\n", internet_users.isnull().sum())
print("Duplicate records:", internet_users.duplicated().sum())

print("\nLITERACY RATE DATASET")
print("Rows:", literacy.shape[0])
print("Columns:", literacy.shape[1])
print("Column names:", literacy.columns.tolist())
print("Data types:\n", literacy.dtypes)
print("Missing values:\n", literacy.isnull().sum())
print("Duplicate records:", literacy.duplicated().sum())
```

**Part C – Data Transformation**

**TASK 6**

```python
gdp = gdp.rename(columns={'Country': 'country'})
literacy = literacy.rename(columns={'name': 'country'})
print("Population columns:", population.columns.tolist())
print("GDP columns:", gdp.columns.tolist())
print("Internet Users columns:", internet_users.columns.tolist())
print("Literacy Rate columns:", literacy.columns.tolist())
```

**TASK 7**

```python
print("POPULATION DATATYPES")
print(population.dtypes)
print('\n')
print("GDP DATATYPES")
print(gdp.dtypes)
print('\n')
print("INTERNET DATATYPES")
print(internet_users.dtypes)
print('\n')
print("LITERACY DATATYPES")
print(literacy.dtypes)
```

**TASK 8**

```python
df1 = pd.merge(population, gdp, on='country', how='outer')
df2 = pd.merge(df1, internet_users, on='country', how='outer')
final_df = pd.merge(df2, literacy, on='country', how='outer')

final_df
```

**TASK 9**

```python
final_df['Internet Penetration (%)'] = round((final_df['internet_users'] / final_df['population']) * 100, 2)
final_df['GDP Per Capita'] = round(final_df['GDP'] / final_df['population'], 2)
final_df
```

**Part D – Load Data**

**TASK 10**

```python
import sqlite3
conn = sqlite3.connect('country_data.db')
final_df.to_sql('country_statistics', conn, if_exists='replace', index=False)
```

**TASK 11**

**Verify the data has been loaded successfully by reading it back from the database**

```python
query = "SELECT * FROM country_statistics"
verification_df = pd.read_sql(query, conn)

verification_df
```

**TASK 12**

**Retrieve the first five records from the SQLite table**

```python
query_five = "SELECT * FROM country_statistics LIMIT 5"
df_five = pd.read_sql(query_five, conn)
display(df_five)
```

**Part F – Data Analysis Using SQL**

**11 Display the Top 5 countries by GDP.**

```python
query_11 = """SELECT country, GDP FROM country_statistics ORDER BY GDP DESC LIMIT 5"""
df_11 = pd.read_sql(query_11, conn)
display(df_11)
```

**12 Find the average literacy rate.**

```python
query_12 = """SELECT AVG(literacy_rate) AS average_literacy_rate FROM country_statistics"""
df_12 = pd.read_sql(query_12, conn)
display(df_12)
```

**13 Display all countries having literacy rate above 90%.**

```python
query_13 = """SELECT country, literacy_rate FROM country_statistics WHERE literacy_rate > 90"""
df_13 = pd.read_sql(query_13, conn)
display(df_13)
```

**14 Find the Top 5 countries with the highest internet penetration.**

```python
query_14 = """SELECT country, `Internet Penetration (%)` FROM country_statistics ORDER BY `Internet Penetration (%)` DESC LIMIT 5"""
df_14 = pd.read_sql(query_14, conn)
display(df_14)
```

**15 Calculate the total population.**

```python
query_15 = """SELECT SUM(population) AS total_population FROM country_statistics"""
df_15 = pd.read_sql(query_15, conn)
display(df_15)
```

**16 Find the country with the highest GDP per capita.**

```python
query_16 = """SELECT country, `GDP Per Capita` FROM country_statistics ORDER BY `GDP Per Capita` DESC LIMIT 1"""
df_16 = pd.read_sql(query_16, conn)
display(df_16)
```
